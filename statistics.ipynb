```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats
```

```python
num_wins_nich = 0
num_moves_nich = []
for i in range(1000):
    deck.shuffle()
    board=deck.deal()
    for j in range(1000):
        board.greed_play(strategy=nich_strategy)
        if board.check_win(): 
            num_wins_nich = num_wins_nich + 1
            num_moves_nich.append(j)
            break
        if board.no_moves:
            break
print(num_wins_nich)

num_wins_yan = 0
num_moves_yan = []
for i in range(1000):
    deck.shuffle()
    board=deck.deal()
    for j in range(1000):
        board.greed_play(strategy=yan_strategy)
        if board.check_win(): 
            num_wins_yan = num_wins_yan + 1
            num_moves_yan.append(j)
            break
        if board.no_moves:
            break
print(num_wins_yan)
```

```python
win_rate_yan = num_wins_yan/1000
win_rate_nich = num_wins_nich/1000
titles = np.array(["Yan et al. 2005", "Our Heuristic"])
win_rates = np.array([win_rate_yan,win_rate_nich])
plt.bar(titles,win_rates)
plt.ylabel("Win Rate")
plt.ylim(0,1)
plt.show()
```

```python
plt.hist(num_moves_nich,bins=30)
plt.ylabel("Frequency")
plt.xlabel("Number of Moves to Win")
plt.show()

plt.hist(num_moves_yan,bins=30)
plt.ylabel("Frequency")
plt.xlabel("Number of Moves to Win")
plt.show()
```

```python
mean_nich = np.mean(num_moves_nich)
mean_yan = np.mean(num_moves_yan)
serr_nich = stats.sem(num_moves_nich)
serr_yan = stats.sem(num_moves_yan)

mean_vect = np.array([mean_yan,mean_nich])
serr_vect = np.array([serr_yan,serr_nich])
plt.bar(titles,mean_vect,width=0.5)
plt.errorbar(titles,mean_vect, yerr=serr_vect, ecolor="black",fmt="none")
plt.ylabel("Average Number of Moves to Win")
plt.show()
```
