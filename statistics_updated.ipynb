```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats
```

```python
# perform 1000 plays for each method and find number of wins and number of moves for each win
num_wins_nich = 0 # number of total wins for our heuristic
num_moves_nich = [] # stores number of moves for each win for our heuristic
for i in range(1000):
    deck.shuffle()
    board=deck.deal()
    for j in range(1000):
        board.greed_play(strategy=nich_strategy)
        if board.check_win(): 
            num_wins_nich = num_wins_nich + 1
            num_moves_nich.append(j)
            break
        if board.no_moves:
            break
print(num_wins_nich)

num_wins_yan = 0 # number of total wins for Yan et al. 2005 method
num_moves_yan = [] # stores number of moves for each win for Yan et al. 2005 method
for i in range(1000):
    deck.shuffle()
    board=deck.deal()
    for j in range(1000):
        board.greed_play(strategy=yan_strategy)
        if board.check_win(): 
            num_wins_yan = num_wins_yan + 1
            num_moves_yan.append(j)
            break
        if board.no_moves:
            break
print(num_wins_yan)
```

```python
# bar graph with win rates for each method
win_rate_yan = num_wins_yan/1000
win_rate_nich = num_wins_nich/1000
titles = np.array(["Yan et al. 2005", "Our Heuristic"])
win_rates = np.array([win_rate_yan,win_rate_nich])
plt.bar(titles,win_rates)
plt.ylabel("Win Rate")
plt.ylim(0,1)
plt.show()
```

```python
# histograms with distribution of number of moves to win
plt.hist(num_moves_nich,bins=50)
plt.ylabel("Frequency")
plt.xlabel("Number of Moves to Win")
plt.title("Our Heuristic")
plt.show()

plt.hist(num_moves_yan,bins=50)
plt.ylabel("Frequency")
plt.xlabel("Number of Moves to Win")
plt.title("Yan et al. 2005")
plt.show()
```

```python
# bar graph with mean number of moves to win for each method
# bars show standard error
mean_nich = np.mean(num_moves_nich)
mean_yan = np.mean(num_moves_yan)
serr_nich = stats.sem(num_moves_nich)
serr_yan = stats.sem(num_moves_yan)

mean_vect = np.array([mean_yan,mean_nich])
serr_vect = np.array([serr_yan,serr_nich])
plt.bar(titles,mean_vect,width=0.5)
plt.errorbar(titles,mean_vect, yerr=serr_vect, ecolor="black",fmt="none")
plt.ylabel("Average Number of Moves to Win")
plt.show()
```
